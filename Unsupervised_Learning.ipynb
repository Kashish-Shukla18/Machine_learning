```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X,y_true = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42)
scaler = StandardScaler()
X
```

```python
df=pd.DataFrame(X,columns=['Feature 1','Feature 2'])
```

```python
df
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
```

```python
wcss=[]
K_range=range(1,11)
for k in K_range:
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(K_range,wcss,marker='o')
```

```python
kmeans_final=KMeans(n_clusters=3,random_state=42)
cluster_labels=kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster']=cluster_labels
```

```python
sns.scatterplot(x=df['Feature 1'],y=df['Feature 2'],hue=df['cluster'],palette='viridis')
```

**DBSCAN**(Density based spatial clustering Alfgo)

```python
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN,KMeans
```

```python
X,y_true=make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python
df=pd.DataFrame(X,columns=['Feature 1','Feature 2'])
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
```

```python
kmeans=KMeans(n_clusters=2,random_state=42)
kmeans_labels=kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster']=kmeans_labels
```

```python
sns.scatterplot(x=df['Feature 1'],y=df['Feature 2'],hue=df['kmeans_cluster'],palette='viridis')
```

```python
dbscan=DBSCAN(eps=0.3,min_samples=5)
dbscan_labels=dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster']=dbscan_labels
```

```python
sns.scatterplot(x=df['Feature 1'],y=df['Feature 2'],hue=df['dbscan_cluster'],palette='tab10')
```
